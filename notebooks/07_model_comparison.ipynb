```python
import pandas as pd
import numpy as np
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        0.735, 
        0.754, 
        0.764, 
        0.804
    ],
    "Precision": [
        0.501, 
        0.526, 
        0.539, 
        0.674
    ],
    "Recall": [
        0.791, 
        0.751, 
        0.749, 
        0.508
    ],
    "F1": [
        0.613, 
        0.619, 
        0.627, 
        0.579
    ],
    "ROC_AUC": [
        0.841, 
        0.837, 
        0.843, 
        0.848
    ]
})

results
```

```python
weights = {
    "ROC_AUC": 0.40,
    "Recall": 0.30,
    "F1": 0.20,
    "Precision": 0.10
}
```

```python
results["Weighted_Score"] = (
    results["ROC_AUC"] * weights["ROC_AUC"] +
    results["Recall"] * weights["Recall"] +
    results["F1"] * weights["F1"] +
    results["Precision"] * weights["Precision"]
)

results.sort_values(by="Weighted_Score", ascending=False)
```

```python
results[["Model", "Weighted_Score"]].sort_values(
    by="Weighted_Score", ascending=False
)
```
