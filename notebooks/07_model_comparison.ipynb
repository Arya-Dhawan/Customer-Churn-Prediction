```python
import pandas as pd
import numpy as np
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        0.734, 
        0.724, 
        0.766, 
        0.794
    ],
    "Precision": [
        0.499, 
        0.487, 
        0.541, 
        0.646
    ],
    "Recall": [
        0.789, 
        0.775, 
        0.770, 
        0.497
    ],
    "F1": [
        0.611, 
        0.598, 
        0.636, 
        0.562
    ],
    "ROC_AUC": [
        0.841, 
        0.827, 
        0.844, 
        0.846
    ]
})

results
```

```python
weights = {
    "ROC_AUC": 0.40,
    "Recall": 0.30,
    "F1": 0.20,
    "Precision": 0.10
}
```

```python
results["Weighted_Score"] = (
    results["ROC_AUC"] * weights["ROC_AUC"] +
    results["Recall"] * weights["Recall"] +
    results["F1"] * weights["F1"] +
    results["Precision"] * weights["Precision"]
)

results.sort_values(by="Weighted_Score", ascending=False)
```

```python
results[["Model", "Weighted_Score"]].sort_values(
    by="Weighted_Score", ascending=False
)
```
